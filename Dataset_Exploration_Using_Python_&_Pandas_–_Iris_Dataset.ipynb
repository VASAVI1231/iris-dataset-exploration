```python
# ============================================
# TASK 17 - DATASET EXPLORATION
# Iris Dataset using Python and Pandas
# ============================================

# 1. Import required libraries
import pandas as pd
from sklearn.datasets import load_iris

print("Libraries imported successfully!")


# ============================================
# 2. Load Iris Dataset
# ============================================

iris = load_iris()

# Create DataFrame
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

# Add target/species column
df["species"] = [iris.target_names[i] for i in iris.target]

print("\nIris dataset loaded successfully!")


# ============================================
# 3. Display Complete Dataset
# ============================================

print("\n========== DATASET ==========")
display(df)


# ============================================
# 4. Dataset Shape
# ============================================

print("\n========== DATASET SHAPE ==========")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
print("Shape:", df.shape)


# ============================================
# 5. Column Names
# ============================================

print("\n========== COLUMN NAMES ==========")

for column in df.columns:
    print(column)


# ============================================
# 6. Dataset Information
# ============================================

print("\n========== DATASET INFORMATION ==========")
df.info()


# ============================================
# 7. Statistical Description
# ============================================

print("\n========== STATISTICAL SUMMARY ==========")
display(df.describe())


# ============================================
# 8. First 5 Rows
# ============================================

print("\n========== FIRST 5 ROWS ==========")
display(df.head())


# ============================================
# 9. Last 5 Rows
# ============================================

print("\n========== LAST 5 ROWS ==========")
display(df.tail())


# ============================================
# 10. Check Missing Values
# ============================================

print("\n========== MISSING VALUES ==========")

missing_values = df.isnull().sum()

print(missing_values)


# ============================================
# 11. Check Duplicate Rows
# ============================================

print("\n========== DUPLICATE ROWS ==========")

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# ============================================
# 12. Identify Numerical Columns
# ============================================

print("\n========== NUMERICAL COLUMNS ==========")

numerical_columns = df.select_dtypes(
    include="number"
).columns

for column in numerical_columns:
    print(column)


# ============================================
# 13. Identify Categorical Columns
# ============================================

print("\n========== CATEGORICAL COLUMNS ==========")

categorical_columns = df.select_dtypes(
    exclude="number"
).columns

for column in categorical_columns:
    print(column)


# ============================================
# 14. Species Distribution
# ============================================

print("\n========== SPECIES DISTRIBUTION ==========")

species_count = df["species"].value_counts()

print(species_count)


# ============================================
# 15. Unique Species
# ============================================

print("\n========== UNIQUE SPECIES ==========")

print(df["species"].unique())


# ============================================
# 16. Number of Unique Species
# ============================================

print("\n========== NUMBER OF SPECIES ==========")

print("Number of unique species:",
      df["species"].nunique())


# ============================================
# 17. Mean Values
# ============================================

print("\n========== MEAN VALUES ==========")

print(df[numerical_columns].mean())


# ============================================
# 18. Minimum Values
# ============================================

print("\n========== MINIMUM VALUES ==========")

print(df[numerical_columns].min())


# ============================================
# 19. Maximum Values
# ============================================

print("\n========== MAXIMUM VALUES ==========")

print(df[numerical_columns].max())


# ============================================
# 20. Final Observations
# ============================================

print("\n========== KEY OBSERVATIONS ==========")

print("1. The Iris dataset contains 150 rows and 5 columns.")

print("2. Four columns contain numerical flower measurements.")

print("3. The dataset contains three Iris species:")
print("   - Setosa")
print("   - Versicolor")
print("   - Virginica")

print("4. The original Iris dataset contains no missing values.")

print("5. The dataset contains no duplicate rows.")

print("6. The head() function displays the first five records.")

print("7. The tail() function displays the last five records.")

print("8. The describe() function provides statistical information.")

print("9. Dataset exploration helps us understand the data before modelling.")


# ============================================
# 21. Final Conclusion
# ============================================

print("\n========== CONCLUSION ==========")

print(
    "Dataset exploration is an important step before applying "
    "machine learning models. Using Python and Pandas, we explored "
    "the Iris dataset, checked its dimensions, columns, data types, "
    "statistics, missing values and species distribution."
)

print("\n========== PROJECT COMPLETED ==========")
```
